# Climate Data Download: ISU Climate Data

Download the data from 2015-2025 from the Iowa State University Mesonet Daily Climate Observations.

## Download and Merge Data

```python
import requests
import pandas as pd
import os

base_url = "https://mesonet.agron.iastate.edu/cgi-bin/request/daily.py"
params_template = {
    "network": "IA_ASOS",
    "stations": "_ALL",
    "month1": 1,
    "day1": 1,
    "month2": 12,
    "day2": 31,
    "var": [
        "max_temp_f", "min_temp_f", "max_dewpoint_f", "min_dewpoint_f",
        "precip_in", "avg_wind_speed_kts", "avg_wind_drct", "min_rh",
        "avg_rh", "max_rh", "climo_high_f", "climo_low_f", "climp_precip_in",
        "snow_in", "snowd_in", "min_feel", "avg_feel", "max_feel", "max_wind_speed_kts"
    ],
    "na": "None",
    "format": "csv"
}

output_dir = "../../../../data/tabular/climate/raw"
os.makedirs(output_dir, exist_ok=True)

filepath = os.path.join(output_dir, f"isu-climate.csv")
os.remove(filepath)

print("removed old data (if present)")

for year in range(2015, 2025):
    print(f"Downloading data for {year}...")
    params = params_template.copy()
    params["year1"] = year
    params["year2"] = year

    try:
        response = requests.get(base_url, params=params, timeout=30)
        response.raise_for_status()
        filepath = os.path.join(output_dir, f"climate_{year}.csv")
        with open(filepath, "w", encoding="utf-8") as f:
            f.write(response.text)
        print(f"Saved: {filepath}")
    except Exception as e:
        print(f"Failed to download data for {year}: {e}")

# Merge all yearly CSVs into one
dfs = []
for year in range(1980, 2025):
    filepath = os.path.join(output_dir, f"climate_{year}.csv")
    if os.path.exists(filepath):
        dfs.append(pd.read_csv(filepath))

merged = pd.concat(dfs, ignore_index=True)
merged_path = os.path.join(output_dir, "isu-climate.csv")
merged.to_csv(merged_path, index=False)
print(f"Merged CSV saved: {merged_path}")

for year in range(2015, 2025):
    filepath = os.path.join(output_dir, f"climate_{year}.csv")
    if os.path.exists(filepath):
        os.remove(filepath)
        print(f"Removed: {filepath}")

print("============= DONE =============")
```
